```python
# Criando a coluna de interação com as regras descritas
Base['interaction'] = np.where(
    (Base['recommended'] == True) & (Base['votes_helpful'] == 0), 1,
    np.where(
        (Base['recommended'] == False) & (Base['votes_helpful'] == 0), -1,
        np.where(
            Base['recommended'] == True, 1 + Base['votes_helpful'],
            -1 - Base['votes_helpful']
        )
    )
)
```

```python
# Convertendo 'Categories' e 'Genres' em listas, caso estejam em formato string
Base['Categories'] = Base['Categories'].fillna('').str.split(', ')
Base['Genres'] = Base['Genres'].fillna('').str.split(', ')

# Aplicando One-Hot Encoding para 'Categories'
categories_dummies = Base['Categories'].str.join('|').str.get_dummies()

# Aplicando One-Hot Encoding para 'Genres'
genres_dummies = Base['Genres'].str.join('|').str.get_dummies()

# Concatenando os dummies ao dataframe original
df = pd.concat([Base, categories_dummies, genres_dummies], axis=1)
```

```python
# Definindo novos jogos (sem interações anteriores)
all_games = set(df['app_id'].unique())
interacted_games = set(df[df['recommended'].notna()]['app_id'].unique())
new_games_ids = list(all_games - interacted_games)

# Selecionando as categorias e gêneros que o usuário gosta, com base nas interações positivas
user_id = 76561198062813911  # Exemplo de ID de usuário
user_liked_games = df[(df['author.steamid'] == user_id) & (df['recommended'] == True)]
user_preferred_categories = user_liked_games[categories_dummies.columns].sum()
user_preferred_genres = user_liked_games[genres_dummies.columns].sum()

# Encontrar novos jogos com categorias e gêneros semelhantes
recommended_new_games = df[(df['app_id'].isin(new_games_ids)) & 
                           (df[categories_dummies.columns].dot(user_preferred_categories) > 0) &
                           (df[genres_dummies.columns].dot(user_preferred_genres) > 0)]['app_name'].unique()

print("Novos jogos recomendados para o usuário:", recommended_new_games)
```
