```python
import numpy as np
np.random.seed(42)
n_simul = 10000
n_auto = 100
n_saude = 50
n_prop = 20

mu_auto, sigma_auto = 5000, 1000
mu_saude, sigma_saude = 3000, 500
mu_prop, sigma_prop = 10000, 2000

losses_auto = np.random.normal(mu_auto, sigma_auto, (n_simul, n_auto)).sum(axis=1)
losses_saude = np.random.normal(mu_saude, sigma_saude, (n_simul, n_saude)).sum(axis=1)
losses_prop = np.random.normal(mu_prop, sigma_prop, (n_simul, n_prop)).sum(axis=1)

total_losses = losses_auto + losses_saude + losses_prop

media = np.mean(total_losses)
mediana = np.median(total_losses)
percentil_90 = np.percentile(total_losses, 90)
percentil_10 = np.percentile(total_losses, 10)

print(f"Média das perdas totais:    ${media:,.2f}")
print(f"Mediana das perdas totais:  ${mediana:,.2f}")
print(f"Percentil 90 das perdas:    ${percentil_90:,.2f}")
print(f"Percentil 10 das perdas:    ${percentil_10:,.2f}")
```
