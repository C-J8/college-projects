```python
# Questão 1
import random

class MDP:
    def __init__(self):
        self.states = ['A', 'B', 'C']
        self.actions = ['Esquerda', 'Direita']
        self.transitions = {
            'A': {'Esquerda': 'A', 'Direita': 'B'},
            'B': {'Esquerda': 'A', 'Direita': 'C'},
            'C': {'Esquerda': 'B', 'Direita': 'C'}
        }
        self.rewards = {
            'A': {'Esquerda': 0, 'Direita': 1},
            'B': {'Esquerda': 0, 'Direita': 0},
            'C': {'Esquerda': 0, 'Direita': 10}
        }

    def next_state(self, state, action):
        return self.transitions[state][action]

    def get_reward(self, state, action):
        return self.rewards[state][action]

# Classe do agente que escolhe ações aleatórias
class RandomAgent:
    def __init__(self, actions):
        self.actions = actions

    def choose_action(self):
        return random.choice(self.actions)

# Função principal de execução
def run_simulation(iterations=20):
    env = MDP()
    agent = RandomAgent(env.actions)
    state = 'A'  
    total_reward = 0

    print("Início da simulação:")
    for i in range(iterations):
        action = agent.choose_action()
        reward = env.get_reward(state, action)
        next_state = env.next_state(state, action)
        total_reward += reward

        print(f"Iteração {i+1}: Estado={state}, Ação={action}, Recompensa={reward}, Próximo Estado={next_state}")
        state = next_state

    print(f"\nRecompensa total acumulada: {total_reward}")

# Executar a simulação
if __name__ == "__main__":
    run_simulation(20)
```

```python
# Questão 2
def value_iteration(mdp, gamma=0.9, epsilon=0.01):
    states = mdp.states
    actions = mdp.actions
    V = {s: 0 for s in states} 

    while True:
        delta = 0
        new_V = {}
        for s in states:
            action_values = []
            for a in actions:
                s_prime = mdp.next_state(s, a)
                reward = mdp.get_reward(s, a)
                value = reward + gamma * V[s_prime]
                action_values.append(value)
            best_value = max(action_values)
            new_V[s] = best_value
            delta = max(delta, abs(V[s] - best_value))
        V = new_V
        if delta < epsilon:
            break

    # Política ótima
    policy = {}
    for s in states:
        best_action = None
        best_value = float('-inf')
        for a in actions:
            s_prime = mdp.next_state(s, a)
            reward = mdp.get_reward(s, a)
            value = reward + gamma * V[s_prime]
            if value > best_value:
                best_value = value
                best_action = a
        policy[s] = best_action

    return V, policy

# Executar
if __name__ == "__main__":
    env = MDP()
    V_opt, policy_opt = value_iteration(env)

    print("Função de valor ótima:")
    for state, value in V_opt.items():
        print(f"V({state}) = {value:.2f}")

    print("\nPolítica ótima:")
    for state, action in policy_opt.items():
        print(f"π({state}) = {action}")
```

A transferência de aprendizado em MDPs permite usar o conhecimento adquirido em uma tarefa para acelerar o aprendizado em outra semelhante. Isso pode ser feito reutilizando políticas, funções de valor ou representações dos estados, principalmente quando os ambientes compartilham estruturas parecidas.

Por exemplo, um agente que aprendeu a se mover em um labirinto pode usar esse conhecimento para se adaptar mais rápido a outro labirinto com layout semelhante. Com isso, o agente precisa de menos interações para aprender e pode alcançar boas decisões com mais eficiência.
