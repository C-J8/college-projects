```python
import spacy
import zipfile
import os
from nltk.util import ngrams


# Carrega o modelo de português do spaCy
nlp = spacy.load("pt_core_news_sm")

# Caminho do arquivo zip e pasta para extração
zip_path = 'datasets/DADOS_FOLHADESAOPAULO.zip'
extract_path = 'DADOS_FOLHADESAOPAULO'

# Extrair o conteúdo do zip (se ainda não tiver extraído)
if not os.path.exists(extract_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)

# Lista os arquivos .txt da pasta
all_files = sorted([
    os.path.join(extract_path, f)
    for f in os.listdir(extract_path)
    if f.endswith(".txt")
])

# Processar os 10 primeiros arquivos
for i, file_path in enumerate(all_files[13:15]):
    print(f"\n===== Arquivo {i+1}: {os.path.basename(file_path)} =====")

    with open(file_path, 'r', encoding='UTF-16') as file:
        text = file.read()

    # Processa o texto com spaCy
    doc = nlp(text)

    # Tokenização em sentenças
    sentences = [sent.text for sent in doc.sents]

    # Tokens alfabéticos
    tokens = [token.text for token in doc if token.is_alpha]

    # Stopwords e lematização
    stop_words = nlp.Defaults.stop_words
    stop_words_in_text = [token for token in tokens if token.lower() in stop_words]

    tokens_lemmatized_no_stopwords = [
        token.lemma_.lower()
        for token in doc
        if token.is_alpha and token.text.lower() not in stop_words
    ]

    # Geração de bigramas e trigramas
    bigrams = list(ngrams(tokens_lemmatized_no_stopwords, 2))
    trigrams = list(ngrams(tokens_lemmatized_no_stopwords, 3))

    # Mostrando um resumo dos resultados
    print("Sentenças:", sentences[:2])
    print("Tokens:", tokens[:5])
    print("Stopwords:", stop_words_in_text[:5])
    print("Tokens lematizados:", tokens_lemmatized_no_stopwords[:5])
    print("Bigramas:", bigrams[:3])
    print("Trigramas:", trigrams[:3])
```
